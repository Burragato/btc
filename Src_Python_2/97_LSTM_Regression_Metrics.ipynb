```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
from sklearn.metrics import mean_absolute_percentage_error

def regression_metrics(y_true, y_confronto):
metrics = {}
if isinstance(y_confronto, str): 
        y_confronto = y_true.shift(-1).dropna() 
        y_true = y_true[:-1]

# Calcola le metriche
mae = mean_absolute_error(y_true, y_confronto) 
mse = mean_squared_error(y_true, y_confronto)
rmse = np.sqrt(mse)
r_squared = r2_score(y_true, y_confronto)
mape = mean_absolute_percentage_error(y_true, y_confronto)

# Aggiungi le metriche al dizionario 

metrics['MAPE'] = mape metrics['MAE'] = mae
metrics['MSE'] = mse metrics['RMSE'] = rmse metrics['R_squared'] = r_squared

return metrics






```
