```python
def crear_lags(df, variable, num_lags = 24):

#Creo l'oggetto Dataframe
lags = pd.DataFrame()

#Creo tutti i lags
for cada in range(1,num_lags+1):
lags[variable + '_lag_'+ str(cada)] = df[variable].shift(cada)

#restitusice il dataframe dei lags
return(lags)


def crear_delta_perc(df, variable, num_lags = 24):

#Creo l'oggetto Dataframe
delta_perc = pd.DataFrame()

#Calcolo la differenza %
for cada in range(1,num_lags+1):
delta_perc[variable + '_delta_perc_'+ str(cada)] =
((df[variable].shift(1) - df[variable].shift(cada+1)) / 
df[variable].shift(cada+1))*100

#restitusice il dataframe
return(delta_perc)


def media_movil(df, variable, num_periodos = 24): 
        mm = pd.DataFrame()
for cada in range(2,num_periodos+1):
mm[variable + '_mm_' + str(cada)] = df[variable].shift(1).rolling(cada).mean()

return(mm)



```
