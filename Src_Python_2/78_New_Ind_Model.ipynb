```python
from keras.layers import Dense, Dropout
from keras.models import Sequential
from sklearn.preprocessing import StandardScaler

#input_shape=(X_train.shape[1],)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) 
X_test_scaled = scaler.transform(X_test)

# Definizione di un modello più complesso
model = Sequential()
model.add(Dense(512, activation='relu'))
model.add(Dropout(0.3)) # Aggiungi dropout per prevenire l'overfitting
model.add(Dense(256, activation='relu')) 
model.add(Dropout(0.3)) 
model.add(Dense(128, activation='relu')) 
model.add(Dropout(0.3)) 
model.add(Dense(64, activation='relu')) 
model.add(Dropout(0.3)) 
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.3)) 
model.add(Dense(16, activation='relu')) 
model.add(Dropout(0.3)) 
model.add(Dense(1, activation='linear'))

# Compilazione del modello
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mse'])

# Addestramento del modello
model.fit(X_train_scaled, y_train, epochs=200, batch_size=128, validation_split=0.2)

# Valutazione del modello
loss, mse = model.evaluate(X_test_scaled, y_test) 
print(f'Test Loss: {loss}, Test MSE: {mse}')


```
